# Registration Visualization
Generate overlays and calcium-distance histogram into outputs/plots/.

```python
import subprocess
import sys
from pathlib import Path

ROOT = Path('..').resolve()
if str(ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(ROOT / 'src'))

from predict_cac.evaluation.visualization import save_registration_plots

generate_demo_script = ROOT / 'scripts' / 'generate_demo_data.py'
```

```python
ct_candidate_1 = ROOT / 'data/preprocessed/sample_ct.nii.gz'
ct_candidate_2 = ROOT / 'data/preprocessed/sample_fixed.nii.gz'
ct_image = ct_candidate_1 if ct_candidate_1.exists() else ct_candidate_2
calcium_mask = ROOT / 'data/masks/sample_calcium_mask.nii.gz'
centerline_mask = ROOT / 'outputs/centerlines_warped/sample_centerline_mask.nii.gz'
plot_dir = ROOT / 'outputs/plots'

required = [ct_image, calcium_mask, centerline_mask]
if not all(p.exists() for p in required):
    print('Missing demo inputs. Generating demo data...')
    subprocess.run([sys.executable, str(generate_demo_script)], cwd=str(ROOT), check=True)
    ct_image = ct_candidate_1 if ct_candidate_1.exists() else ct_candidate_2

for p in [ct_image, calcium_mask, centerline_mask]:
    print(p, 'exists=' + str(p.exists()))
```

```python
if ct_image.exists() and calcium_mask.exists() and centerline_mask.exists():
    outputs = save_registration_plots(ct_image, calcium_mask, centerline_mask, plot_dir)
    print('Generated plot files:')
    for k, v in outputs.items():
        print(f' - {k}: {v}')
else:
    raise FileNotFoundError('Required files are still missing after demo generation.')
```
